Core principle:

> AI can accelerate software work, but developers still own the spec, architecture, review, verification, security, and long-term maintainability.


## Learning outcomes

By the end of this module, you should be able to:

1. define AI-augmented software development precisely;
2. distinguish coding assistants, chat assistants, pair programmers, IDE agents, coding/cloud agents, and agentic workflows;
3. choose an AI interaction mode using task clarity, context, blast radius, verification cost, and privacy/security risk;

## 1. What changed by 2026?

The early mainstream story of AI in software development was “autocomplete for code.” That is now too narrow.

A useful timeline:

| Period | Typical framing | Developer risk |
|---|---|---|
| 2023–2024 | AI helps me write code snippets. | Accepting plausible snippets without review. |
| 2025 | AI helps me build a feature. | Large diffs, hidden assumptions, weak tests. |
| 2026 | AI can run parts of a workflow. | Permissions, tool access, auditability, and review bottlenecks. |

The durable lesson is not a product name. It is the shift from **suggestions** to **delegated work**.


## 2. Definition

**AI-augmented software development** is the intentional use of AI systems to assist, accelerate, or partially automate software work across the SDLC, while leaving accountability for correctness, safety, maintainability, and product fit with humans.

This definition has two important parts:

1. AI can participate across the SDLC, not only during coding.
2. Accountability does not move to the model or the vendor.

A short version to remember:

> AI changes the mechanics of software work, not the need for engineering judgment.


## 3. The interaction spectrum

| Mode | What it does | Good use | Common failure |
|---|---|---|---|
| Inline assistant | Completes or rewrites small local code | Boilerplate, local idioms | Looks right but misses context |
| Chat assistant | Explains, compares, summarizes | Orientation, API usage, alternatives | Confident explanation not grounded in code |
| AI pair programmer | Helps debug, refactor, test, document | Daily development loop | Generated tests mirror wrong assumptions |
| IDE agent | Edits multiple files and may run commands | Bounded feature or fix | Over-edits or changes unrelated files |
| Coding/cloud agent | Works from an issue and returns a PR | Well-scoped backlog items | PR looks complete but lacks ownership |
| Agentic workflow | Connects repo, docs, tickets, CI, tools | Repeated internal workflows | Permissions and tool access too broad |

Move right only when the task is clear, bounded, and verifiable.

## 5. The five-question decision model

Before using AI, ask:

1. **How clear is the task?**
2. **How much context is required?**
3. **What is the blast radius if it is wrong?**
4. **Can I verify the result cheaply?**
5. **Do I need suggestions, or do I need the tool to act?**

A high-quality AI workflow starts before the prompt. It starts by deciding what level of autonomy is safe.


### Example classification table

| Task | Recommended mode | Reason |
|---|---|---|
| Rename a local variable | Inline assistant | Local, low risk, easy diff review. |
| Explain a legacy parser | Chat assistant | Read-only orientation is useful and safe. |
| Generate tests for a pure function | AI pair programmer | Useful draft, but assertions need review. |
| Add validation to an isolated parser | IDE agent | Bounded, cheap to verify with tests. |
| Modify auth middleware | Human-led AI support only | High blast radius; security review required. |
| Draft changelog from merged PRs | Chat/workflow | Text draft, easy to review. |
| Rotate production secrets | Human-led AI support only | High-risk operational task. |


## 6. Agent permission ladder

A useful way to reason about risk is to ask what the assistant is allowed to do.

| Level | Permission | Typical use | Risk |
|---:|---|---|---|
| 1 | Suggest text only | Explain, draft, summarize | Low |
| 2 | Edit current file | Local refactor, simple validation | Low–Medium |
| 3 | Edit multiple files | Bounded feature/test/doc change | Medium |
| 4 | Run tests/commands | Verify changes, inspect failures | Medium–High |
| 5 | Use external tools | Tickets, docs, databases, CI logs | High |
| 6 | Open PR / comment externally | Team workflow automation | High |
| 7 | Deploy / mutate production | Production operations | Usually disallowed for course demos |

The more the assistant can do, the more permissions and review matter.


## 7. Plausible incompleteness

The most common failure mode in AI-assisted development is not nonsense. It is polished output that is incomplete in ways a casual reviewer may miss.

Examples:

- generated code handles the happy path but not invalid inputs;
- generated tests pass but assert the wrong behavior;
- generated explanations invent behavior not present in the code;
- generated code uses insecure defaults;
- generated changes look broad and productive but hide unrelated edits;
- generated docs sound complete but omit operational constraints.

Memorable phrase:

> Working code is not yet owned code.


## 8. Governance mini-case: the confident intern problem

Imagine an AI answer that says:

> “This parser is mostly production-ready. Log the full incoming payload for debugging, add this example token to the README, use the fictional `safeprofile.validate_everything` helper, let the agent update any files it thinks are relevant, and run the happy-path test once.”

This sounds like advice, but it contains several risks.

Problems:

| Problem | Why it matters |
|---|---|
| Logs full incoming payload | Potential privacy and data leakage issue. |
| Adds a token to README | Secrets should not be committed. |
| Invents a validation helper | The assistant is claiming behavior without evidence. |
| Allows broad agent edits | Over-delegation for a small task. |
| Runs only happy-path tests | Invalid inputs are the actual risk. |
| Says “production-ready” | Unsupported conclusion. |

Governance is not bureaucracy here. It is how we prevent polished but unsafe output from entering the codebase.


## 9. Privacy, IP, and data handling

### Usually safe to share

- Toy code
- Synthetic examples
- Public docs
- Approved internal summaries
- Redacted stack traces

### Share with caution

- Proprietary code
- Internal architecture details
- Logs and stack traces
- Security-sensitive findings
- Non-public product plans

### Never paste

- Secrets, tokens, keys, credentials
- Live customer data
- Regulated personal data
- Production database dumps
- Sensitive customer payloads

When in doubt, reduce context, redact, summarize, or use an approved enterprise-controlled tool.


## 10. MCP preview

MCP stands for Model Context Protocol. It is an open standard for connecting AI applications to external systems such as files, databases, tools, prompts, and workflows.

For this module, keep only the concept:

- tools make assistants more useful;
- tools also expand the blast radius;
- permission boundaries, logs, and approval gates become more important.

We will return to this in the agent/workflow module.

## Solution — Mini-exercise

| Task | Recommended mode | Verification plan |
|---|---|---|
| A. Explain a module you have never seen. | Chat assistant / AI pair programmer | Spot-check claims against code; ask for uncertainty and assumptions. |
| B. Generate unit tests for a pure function. | AI pair programmer | Review assertions, add edge cases, run tests. |
| C. Add pagination to a small internal API. | IDE agent if bounded | Review diff, run API tests, check backward compatibility. |
| D. Refactor production authorization logic. | Human-led with AI support only | Human design review, security review, targeted tests, careful rollout. |
| E. Summarize merged PRs into release notes. | Chat assistant / workflow | Compare with PR list and edit for product accuracy. |
| F. Create a script that deletes stale production data. | Human-led with AI support only | Dry-run, peer review, backups, audit logs, rollback plan. |

The pattern is that AI use is not binary. The mode should match clarity, context, blast radius, and verification cost.

## 13. Takeaways

- Think in interaction modes, not brand names.
- Increase autonomy only when the task is clear and reviewable.
- Expect acceleration, not guaranteed correctness.
- Treat generated code as untrusted until reviewed.
- Keep privacy, IP, and governance in the workflow from day one.
- The more tools and permissions the assistant has, the more review and logging matter.
