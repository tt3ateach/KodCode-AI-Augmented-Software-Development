## 1. Learning outcomes

By the end of this module, students should be able to:

- define vibe coding in a disciplined developer context;
- distinguish Module 2-style AI assistance from Module 3-style intent-led first drafting;
- use **Brief → Explore → Contain → Generate → Review → Own**;
- write broad Stage A prompts without letting them become uncontrolled delegation;
- identify when vibe coding is a good fit and when it backfires;
- review and repair an AI-generated first draft;
- connect the frontend to the Module 2 Work Items API;
- explain what they own after the AI-generated draft.

## 2. What vibe coding is — and is not

In this course, **vibe coding** means:

> A conversational, natural-language-driven workflow in which the developer starts closer to product intent and lets AI take a larger share of first-draft implementation work than in ordinary assistant-style coding.

It does **not** mean:

- no requirements;
- no tests;
- no review;
- no ownership;
- shipping the first thing the model generates.

A short classroom phrase:

> The vibe is in the first draft. The professionalism is in what happens after.

## 4. The workflow spine

| Stage | Student question | AI role | Human responsibility |
|---|---|---|---|
| Brief | What experience are we trying to create? | Interpret product intent | Choose useful scope |
| Explore | What first version could exist? | Propose options | Select simplest useful direction |
| Contain | What is the smallest safe slice? | Plan files and steps | Reject overreach |
| Generate | Can we produce a working draft? | Implement first slice | Review and verify |
| Review | What is wrong or weak? | Find risks | Inspect code and tests |
| Own | Can we explain and extend it? | Assist cleanup and docs | Take accountability |

## 5. Prompt progression

The safest pattern is three-stage prompting.

### Stage A — Explore product intent

Start broad, but do not edit files yet.

```text
Read docs/design-brief.md.
Propose two useful first-version experiences for this task tracker.
Focus on visible user flow, core behavior, and what should be deliberately omitted from the first draft.
Do not edit files yet.
```

### Stage B — Contain the implementation

```text
Choose the smaller direction.
Before editing, propose the smallest vertical slice.
Include files to touch, files to avoid, responsibilities, stop conditions, tests, and manual browser checks.
Do not edit files yet.
```

### Stage C — Own the result

```text
Review this implementation for handoff risk.
Identify confusing responsibilities, unsafe DOM behavior, weak tests, accessibility issues, and places where visual polish hides maintainability problems.
Do not edit yet.
```

## 5.1 Solution — Stage A opening interaction

A strong Stage A response should offer options similar to these:

### Option 1 — Small local-first sprint board

- render seeded tasks;
- quick-add a task;
- filter by status;
- toggle todo/done;
- persist in localStorage;
- show empty state.

Why it is strong: visible, useful, testable, low integration risk.

### Option 2 — Richer sprint console

- search;
- priority editing;
- detail panel;
- API sync;
- responsive polish;
- maybe generated visual treatments.

Why it is risky as a first version: larger diff, more moving parts, more manual verification, higher chance of hidden fragility.

Accepted classroom decision:

> Start with Option 1, then add minimal API integration during ownership phase.

## 5.2 Solution — Stage B containment plan

Accepted plan:

| File | Responsibility |
|---|---|
| `src/state.js` | Normalize tasks, filter tasks, summarize tasks, translate API status names. |
| `src/localService.js` | localStorage-backed service and safe parse fallback. |
| `src/apiClient.js` | Fetch-based client for Module 2 API. |
| `src/service.js` | Choose local or API mode. |
| `src/render.js` | Safe DOM rendering. |
| `src/app.js` | Event wiring and refresh loop. |
| `tests/state.test.mjs` | Pure state behavior and API mapping. |
| `tests/apiClient.test.mjs` | Fetch URL/method/payload/error tests. |

Stop conditions:

- no build tools;
- no dependencies;
- no dark mode;
- no drag-and-drop;
- no authentication;
- no deployment;
- no rich UI before core behavior works.

## 5.3 Solution — Stage C ownership review

A strong ownership review should find at least some of these risks:

- generated UI may look complete before behavior is verified;
- `innerHTML` is unsafe if used for task titles;
- localStorage parsing needs a fallback for corrupt data;
- status naming differs between frontend `in-progress` and API `in_progress`;
- DOM behavior needs manual browser checks;
- generated tests may cover pure logic but not the actual browser flow;
- API unavailable behavior needs a useful user message;
- first draft may hide too much logic in one file.

Ownership cleanup should be small. A good first cleanup is safe DOM rendering or localStorage corruption handling.

## 6. When vibe coding fits

Good fit:

- greenfield UI slices;
- CRUD-heavy internal tools;
- dashboards and admin panels;
- prototypes and demos;
- low-blast-radius first drafts;
- tasks where review is cheap.

Poor fit:

- authentication and authorization;
- billing or payments;
- production migrations;
- safety-critical flows;
- hard-to-test behavior;
- regulated data handling;
- unclear product requirements.

## 7. Tool forms and Devin-style workflows

A concise distinction:

| Workflow | Human steering | Typical result |
|---|---|---|
| Conversational vibe coding | frequent steering | first draft / visible slice |
| IDE agent mode | bounded local edits | modified files and tests |
| Issue-driven coding agent | less continuous steering | branch, PR, report, or artifact |

“Devin-style” in this module means issue-driven delegation: assign an outcome, let an agent work in an isolated environment, then review a branch/PR/artifact. It is a workflow pattern, not a product requirement.

Key line:

> Less steering does not mean less review. It usually means more review.

## 12.2 Solution — Phase 1 manual browser checks

Manual browser checklist:

- page loads without console errors;
- seeded tasks render;
- quick-add adds a new task;
- blank title shows an error or is rejected;
- status filter updates visible tasks;
- completion toggle changes visual state and summary counts;
- refresh preserves local tasks;
- task title such as `<img src=x onerror=alert(1)>` is displayed as text, not HTML;
- small viewport remains usable.

A good student submission includes both automated tests and this manual checklist.

## 13.1 Solution — Phase 2 ownership cleanup

A strong cleanup candidate is safe DOM rendering.

Problem:

```js
container.innerHTML = tasks.map(task => `<h3>${task.title}</h3>`).join("");
```

This can render user-provided text as HTML. It also encourages a monolithic render string that is harder to review.

Better:

```js
const title = document.createElement("h3");
title.className = "task-title";
title.textContent = task.title;
card.append(title);
```

Why this is ownership work:

- it reduces security risk;
- it makes the DOM structure easier to inspect;
- it separates data from markup;
- it is a small, reviewable change.

## 13.2 Solution — Module 2 API integration

Minimal required integration:

- `GET /work-items` → render tasks;
- `POST /work-items` → quick-add;
- `PATCH /work-items/{id}` with status → toggle done;
- `PATCH /work-items/{id}` with priority → priority change;
- show error when API unavailable;
- keep local mode as fallback.

Important mapping:

| Frontend | Module 2 API |
|---|---|
| `in-progress` | `in_progress` |
| `owner` | `assignee` |
| `createdAt` | `created_at` |
| `updatedAt` | `updated_at` |

A strong solution has API-client tests, not only manual browser checks.

## 14.1 Solution — flawed generated draft review

The flawed draft in `review_exercises/flawed_generated_draft/` is intentionally problematic.

Expected findings:

| Finding | Why it matters |
|---|---|
| `innerHTML` with task title | Unsafe rendering of user-controlled data. |
| Global `window.toggle` | Leaky structure and harder testing. |
| Unguarded `JSON.parse` | Corrupt localStorage breaks the app. |
| Monolithic file | Harder to review and own. |
| No tests | Visual behavior is not proof. |
| Inline event handlers | Harder to reason about and maintain. |
| No API error handling | End-to-end integration will fail poorly. |
| Looks simple and polished | Aesthetic simplicity can hide engineering risk. |

## 15. Takeaways

- Vibe coding is useful for first drafts and exploration.
- Broad first prompts must quickly become bounded plans.
- Visual polish is not ownership.
- Existing codebases require read-only orientation before broad generation.
- Module 2 verification habits become more important in Module 3.
- End-to-end integration must be small, explicit, and reviewed.
- The real skill is knowing where natural-language generation ends and engineering ownership begins.
